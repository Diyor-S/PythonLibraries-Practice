```python
from datetime import datetime, timedelta, timezone
from sqlalchemy import DateTime, create_engine
from sqlalchemy.orm import DeclarativeBase, Mapped, mapped_column, sessionmaker
```

```python
class Base(DeclarativeBase):
    pass
```

```python
class User(Base):
    __tablename__ = "users"

    __table_args__ = {
        "extend_existing": True
    }

    _id: Mapped[int] = mapped_column(primary_key=True)
    created_at: Mapped[datetime] = mapped_column(
        DateTime(timezone=True), default=lambda: datetime.now(timezone.utc)
    )
    expires_at: Mapped[datetime] = mapped_column(
        DateTime(timezone=True),
        default=lambda: datetime.now(timezone.utc) + timedelta(days=1),
    )

    @property
    def is_expired(self) -> bool:
    # Safe check in case expires_at hasn't been populated by flush yet
        if self.expires_at is None:
            return False
        return datetime.now(timezone.utc) >= self.expires_at
```

```python
engine = create_engine("sqlite:///:memory:", echo=False)
SessionLocal = sessionmaker(bind=engine)
```

```python
Base.metadata.create_all(engine)
```

```python
session = SessionLocal()
```

```python
# Create instance (defaults are not in __dict__ yet)
user = User()
print("1. Before session.add() & flush():")
print(f"   created_at: {user.created_at}")
print(f"   expires_at: {user.expires_at}")

# Add to session and flush to trigger defaults
session.add(user)
session.flush()

print("\n2. After session.flush():")
print(f"   created_at: {user.created_at}")
print(f"   expires_at: {user.expires_at}")
print(f"   is_expired: {user.is_expired}")

# Clean up session
session.close()
```
